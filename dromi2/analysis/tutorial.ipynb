```python
# read binary files into numpy

def read_3d_array_from_binary(filename):
    # Open the binary file in read mode
    with open(filename, 'rb') as f:
        # Read the dimensions (x, y, z) of the array
        x_dim, y_dim, z_dim = struct.unpack('3i', f.read(12))

        # Calculate the total number of elements in the array
        total_elements = x_dim * y_dim * z_dim

        # Read the data (flattened array) from the binary file
        data = np.fromfile(f, dtype=np.float32, count=total_elements)

        # Reshape the data into the 3D array with the given dimensions
        array = data.reshape((x_dim, y_dim, z_dim))

    return array


def read_matrix_from_binary(filename):
    # Open the binary file in read mode
    with open(filename, 'rb') as f:
        # Read the dimensions (x_dim, y_dim) of the matrix
        x_dim = np.fromfile(f, dtype=np.int32, count=1)[0]
        y_dim = np.fromfile(f, dtype=np.int32, count=1)[0]

        # Calculate the total number of elements in the matrix
        total_elements = x_dim * y_dim

        # Read the matrix data (1D array)
        data = np.fromfile(f, dtype=np.float32, count=total_elements)

        # Reshape the data into the 2D matrix with the given dimensions
        matrix = data.reshape((x_dim, y_dim))

    return matrix
```
